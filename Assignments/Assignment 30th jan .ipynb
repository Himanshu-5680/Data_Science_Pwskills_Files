1. Write a program to accept percentage from the user and display the grade according to the following
criteria:
Marks Grade
    >90 A
    >80 and >=60 B
    >=60 and <=80 C
    below 60 D

```python
percentage=float(input("Please enter % in numbers only."))
print("-" * 40) # एक सुंदर डिवाइडर लाइन के लिए
if percentage > 90:
    print(f"🎉 Congratulations! You scored {percentage}%. Grade: [A]")
    print("🌟 Outstanding performance! Keep shining!")
elif percentage > 80 and percentage <= 90:
    print(f"👏 Great Job! You scored {percentage}%. Grade: [B]")
    print("🚀 Excellent work! You are doing amazing!")
elif percentage >= 60 and percentage <= 80:
    print(f"👍 Good Effort! You scored {percentage}%. Grade: [C]")
    print("📈 Good progress! Aim higher next time!")
else:
    print(f"📚 Keep Learning! You scored {percentage}%. Grade: [D]")
    print("💪 Don't give up! Hard work always pays off.")
print("-" * 40) # एक सुंदर डिवाइडर लाइन के लिए
```

2.Write a program to accept the cost price of a bike and display the road tax to be paid according to the
following criteria:
Tax Cost Price(in Rs)
15%
10%
5%
>100000
>50000 and <= 100000
< = 50000

```python
price_of_bike = float(input("🏍️ Enter the Cost Price of the bike (in Rs.): "))
print('-'  * 40)
if price_of_bike >100000:
    tax_cost = 15
    tax_amount = (price_of_bike*15)/100
    print(f"📊 Applied Road Tax Rate: [15%]")
    
elif price_of_bike > 50000 and price_of_bike <=100000:
    tax_cost = 10
    tax_amount = (price_of_bike*10)/100
    print(f"📊 Applied Road Tax Rate: [10%]")
else :
    tax_cost = 5
    tax_amount = (price_of_bike*5)/100
    print(f"📊 Applied Road Tax Rate: [5%]")
    
# on road price calculation.
total_price = tax_amount + price_of_bike

print(f"💰 Base Cost Price    : ₹{price_of_bike:,.2f}")
print(f"💸 Road Tax to be Paid: ₹{tax_amount:,.2f}")
print("-" * 45)
print(f"🏁 Final On-Road Price: ₹{total_price:,.2f}")
print("-" * 45)
print("🚀 Ride safe! Always wear a helmet. 🪖")
```

```python
city = input( ("🏙️ Please enter the name of a city ( Delhi / Agra / Jaipur )")).strip().title()

print("-" * 50)

if city == "Delhi":
    print(f"🏰 Welcome to the Capital City, {city}!")
    print(f"✨ Famous Monument to visit: [ Red Fort (लाल किला) ]")
    print("ℹ️ Tip: Don't miss the amazing light and sound show in the evening!")
    
elif city == "Agra":
    print(f"🕌 Welcome to the City of Love, {city}!")
    print(f"✨ Famous Monument to visit: [ Taj Mahal (ताज महल) ]")
    print("ℹ️ Tip: One of the 7 Wonders of the World! Early morning is the best time to visit.")
    
elif city == "Jaipur":
    print(f"🌸 Welcome to the Pink City, {city}!")
    print(f"✨ Famous Monument to visit: [ Jal Mahal (जल महल) ]")
    print("ℹ️ Tip: Located in the middle of the Man Sagar Lake, it looks magical at sunset!")
    
else:
    print(f"🗺️ Sorry! Currently, our database doesn't have details for '{city}'.")
    print("📌 Please try searching for Delhi / Agra / or Jaipur.")

print("-" * 50)
```

```python
# 6. Use nested while loop to print 3 different pattern.
```

```python
# Making pattens using nested while loop.
x = 1
print("*"*5 ,'|',"Pattern NO - 1 " '= ' ,end='')
print("Triangle" '|', "*"*5 )
while x <= 5:
    y = 1
    while y <= x:
        print("|☺️|",end = '')
        y+=1    
    x+=1
    print()
```

```python
# Making pattens using nested while loop.
x = 5 
print("*"*5 ,'|',"Pattern NO - 2 " '= ' ,end='')
print("Reverse Triangle" '|' ,"*"*5 )
while x>=1:
    y = 1
    while y<=x:
        print('|☺️|' , end = '')
        y+=1 
    x-=1
    print()
```
