```python
# add(x) adding a new user.

active_users ={101,102}
active_users.add(103)
print(active_users)

# update() : दूसरे सर्वर का डेटा एक साथ जोड़ना

new_users = [104,105]
active_users.update(new_users)
print(active_users)
```

# ---------------------------------------------------------
# 3. आइटम हटाना (Removing Elements)
# ---------------------------------------------------------

```python
# remove(x) : रिज़ॉल्व हुए बग का ID हटाना (Error देगा अगर ID नहीं है)
bugs = {'b1','b2','b3'}
bugs.remove('b2')
print(bugs)

# discard(x) : न्यूज़लेटर लिस्ट से अनसब्सक्राइब करना (Error नहीं देगा)
emails = {'a@x.com','b@x.com'}
emails.discard('c@x.com')
print(emails)

# pop() : रैंडम पेंडिंग टास्क उठाना और हटाना
tasks = {'task1','task2','task3'}
print(tasks.pop())
print(tasks)

# clear() : आज के टेम्परेरी सेशंस पूरी तरह खाली करना
sessions = {'s1','s2'}
sessions.clear()
print(sessions)
```

# ---------------------------------------------------------
# 4. गणितीय सेट ऑपरेशन्स (Mathematical Set Operations)
# ---------------------------------------------------------

```python
web_users = {1, 2, 3}
app_users = {3, 4, 5}

# union (|) : tatal audience
print(web_users | app_users)

# intersetion (&) : Common users jo dono chalate hai.
print(web_users & app_users)

# difference (-) : only web users
print(web_users - app_users)

# symmetric_difference (^) : jo kisi ek ka use karte hai.
print(web_users ^ app_users) 
# common users ko remove kar deta hai.
```

# ---------------------------------------------------------
# 5. सेट अपडेट ऑपरेशन्स (In-place Updates)
# ---------------------------------------------------------

```python
# isdisjoint() A/B टेस्टिंग में चेक करना कि टेस्ट और कंट्रोल ग्रुप अलग हैं ना?
test_group = {1,2}
control_group = {3,4}
print(test_group.isdisjoint(control_group))

# issubset (<=) क्या यूज़र की परमिशन, एडमिन की परमिशंस का हिस्सा हैं?
user_perm = {'read'}
admin_perm = {'read','write','delete'}
print(user_perm <= admin_perm)
print(user_perm < admin_perm) # proper subset

#issuperset (>=) : क्या मास्टर डेटाबेस में सारे चेक किए गए IDs मौजूद हैं?
master_db = {1,2,3,4,5}
check_ids = {2,3}
print(master_db >= check_ids)
print(master_db > check_ids) # proper superset.
```

# 7. कॉपी और एडवांस्ड (Copy & Advanced)
# ---------------------------------------------------------

```python
# copy() : मॉडिफाई करने से पहले डेटाबेस का बैकअप लेना
original_data = {1,2,3}
backup_data = original_data.copy()
print(backup_data)

# frozenset() : फिक्स कैटेगरीज (जिन्हें कोई बदल न सके) 
categories = frozenset({'Electronics','Clothing','Books'})
print(categories)

# Set Comprehension : लाखों प्राइस डेटा से यूनिक डिस्काउंटेड प्राइस बनाना
prices = [100,200,100,300]
discount_unique = {p*0.9 for p in prices}
print(discount_unique)

# max() / min() / sum() : सेल्स डेटा एनालिसिस
sales  = {1500,2500,3500}
print(max(sales))
print(min(sales))
print(sum(sales))

# any() / all() : सर्वर स्टेटस चेक (0 मतलब डाउन, 1 मतलब अप)
server_status = {1,1,0}
print(any(server_status)) # क्या कोई एक भी सर्वर चालू है?
print(all(server_status)) # क्या सारे सर्वर चालू हैं?

# enumerate() : सेट के आइटम्स को इंडेक्स देना (लूप में)
status_codes = {'OK',"ERROR","PENDING"}
print(list(enumerate(status_codes)))
```

* 2. Mutual Fund Overlap (Quant Analysis)


* समस्या: एक इन्वेस्टर दो पैसिव म्यूचुअल फंड्स का डेटा एनालाइज कर रहा है। ऐसे स्टॉक्स निकालो जो 'Fund A' और 'Fund B' दोनों में कॉमन हैं, लेकिन वो 'high_risk' लिस्ट में बिल्कुल नहीं होने चाहिए (Chaining Set Operations)।
* सैंपल डेटा:





Python

`fund_a_stocks = {"Reliance", "TCS", "HDFC", "Zomato", "Infosys"}`
`fund_b_stocks = {"HDFC", "Zomato", "ITC", "Reliance", "SBI"}`
`high_risk_stocks = {"Zomato", "Paytm", "YesBank"}`


```python
fund_a_stocks = {"Reliance", "TCS", "HDFC", "Zomato", "Infosys"}
fund_b_stocks = {"HDFC", "Zomato", "ITC", "Reliance", "SBI"}
high_risk_stocks = {"Zomato", "Paytm", "YesBank"}

passive_stocks = (fund_a_stocks & fund_b_stocks)-high_risk_stocks
print(passive_stocks)
```
