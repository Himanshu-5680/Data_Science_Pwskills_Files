## 2. डेटा निकालना (Accessing Elements) - ⭐️ [Very Important]

```python
model_metrics = {'accuracy':0.85 , 'precision':0.82
                , 'recall':0.88}

print(model_metrics['recall'])

# उदाहरण 1: get() का यूज़ (Missing Data Handle करना)
f0 = model_metrics.get('accuracy')
print(f0)
# अगर f1_score नहीं है, तो Error की जगह 0.0 दे दो
f1 = model_metrics.get('f1_score',0.0)
print(f1)

# कैटेगोरिकल फीचर एनकोडिंग में
city_codes = {"Delhi":1,"Mumbai":2}
print(city_codes.get("Pune",-1)) # -1 for unknown category.

# फ्रीक्वेंसी/काउंटिंग डिक्शनरी में
word_counts = {"data":5 , "AI":2}
word = "Machine"
word_counts[word] =  word_counts.get(word,0)+1 # if word nhi hai to starts with 0.
print(word_counts[word])

# उदाहरण 2: keys() का यूज़ (Dataset के सभी कॉलम नाम निकालना)
features = list(model_metrics.keys())
print(features)

# उदाहरण 3: values() का यूज़ (सभी स्कोर्स का औसत/Mean निकालना)
avg_score = sum(model_metrics.values()) / len(model_metrics)
print(avg_score)

# उदाहरण 4: items() का यूज़ (रिपोर्ट प्रिंट करना)
for metric , score in model_metrics.items():
    print("Model" , metric , "is" , score*100 , "%")
```

## 3. डेटा जोड़ना और अपडेट करना (Adding & Updating) - ⭐️ [Very Important]

```python
# उदाहरण 1: update() - दो अलग-अलग सोर्स से आए डेटा को मिलाना 
base_info = {'id':1 , 'product':"Laptop"}
shipping_info = {'status':"Delivered","data":"15-Aug"}
base_info.update(shipping_info)
print(base_info)

# उदाहरण 1: दूसरी डिक्शनरी से अपडेट करना
params = {"lr": 0.1, "loss": "mse"}
new_params = {"lr": 0.05, "batch": 64}
params.update(new_params)
print(params)

# उदाहरण 2: Key-Value पेयर्स (Tuples) से अपडेट करना
params.update([("epochs",50),("loss","mae")])
print(params)

# उदाहरण 3: डायरेक्ट Arguments पास करके अपडेट करना
params.update(optimizer="sgd",metrics=["accuracy"])
print(params)

# उदाहरण 4: मल्टीपल API रिस्पॉन्स को एक मास्टर डिक्शनरी में मर्ज करना
master_record = {}
api_call_1 = {"user_1":"clicked"}
api_call_2 = {"user_2":"bounced"}
master_record.update(api_call_1)
master_record.update(api_call_2)
print(master_record)

# setdefault(key, default) : अगर Key है तो वैल्यू देगा, नहीं है तो
# डिक्शनरी में नई Key-Value जोड़ देगा
# उदाहरण 2: setdefault() - कैटेगरी के हिसाब से डेटा ग्रुप करना (Grouping)
# अगर 'Tech' की मौजूद नहीं है, तो पहले खाली लिस्ट [] बनाओ, फिर डेटा डालो

category_items = {}
category_items.setdefault('Tech' , []).append("Mobile")
category_items.setdefault('Tech',[]).append("Tablet")
print(category_items)

# उदाहरण 3: Merge Operator (|) - (Python 3.9+) डिक्शनरी को आसानी से मर्ज करना

default_configs = {"learning_rate": 0.01, "epochs": 50}
user_configs = {"epochs": 100, "batch_size": 32}
# user_configs वाला डेटा ओवरराइड कर देगा
final_model_configs = default_configs | user_configs
print(final_model_configs)

dict1={"a":1,"b":2} 
dict2={"b":3,"c":4}
merged = dict1 | dict2
dict1 |= dict2 # dict1 ko update kar dega.
dict1
```

### 5.डिक्शनरी व्यूज़ और लूपिंग (Views & Iteration) - [IMPORTANT]
---

```python
metrics = {"accuracy":0.92,"recall":0.85,"precision":0.88}

# keys(): keys ki list.
print(metrics.keys())

# values(): values ki list. 
print(metrics.values())

# items() : key and value dono ko ek saath dena (looping ke through)
# उदाहरण 1: मॉडल के रिज़ल्ट्स प्रिंट करना
for metrics_name , score in metrics.items():
    print(f"{metrics_name.capitalize()}: {score*100}%")
    
# उदाहरण 2: डेटा क्लीनिंग (Null values खोजना)
raw_data = {"age":25 , "income":None , "score":80}
missing_cols = [col for col,val in raw_data.items() if val is None]
print(missing_cols)

#उदाहरण 3: फीचर स्केलिंग (किसी खास फीचर की वैल्यू को 10 से भाग देना)
for feature , val in metrics.items():
    metrics[feature] = val / 10 
print(metrics) 

# उदाहरण 4: डिक्शनरी की Keys और Values को आपस में बदलना 
# (Reverse Mapping)
id_to_name = {1:"cat",2:"Dog"}
name_to_id = {name:id for id , name in id_to_name.items()}
print(name_to_id)
```

### 7. Dictionary Comprehension - ⭐️ [Data Scientist's Favorite]
---

```python
sales = {"mon":100,"tue":200,"wed":50,"thu":300}

# उदाहरण 1: ट्रांसफॉर्मेशन (सभी सेल्स को डॉलर से रुपये में बदलना)
sales_inr = {day:amount*95 for day , amount in sales.items()}
print(sales_inr)

# उदाहरण 2: फिल्टरिंग (सिर्फ उन दिनों का डेटा रखना जहाँ सेल 100 से ज़्यादा हो)
high_sales = {day:amount for day , amount in sales.items() if amount > 100}
print(high_sales)

# उदाहरण 3: फीचर एनकोडिंग (Categorical से Numeric में बदलना)
categories = ["Low","Medium","High"]
category_map = {cat: i for i,cat in enumerate(categories)}
print(category_map)
```

### 8. Dictinoary comprehension with Data Transformation.
---

```python
# उदाहरण 1: किसी लिस्ट की फ्रीक्वेंसी डिक्शनरी बनाना (Word count)
words = ["yes","no","yes","no","yes","no"]
words_freq = {w:words.count(w)for w in set(words)}
print(words_freq)

# उदाहरण 2: डिक्शनरी फ़िल्टर करना 
# (सिर्फ वो फीचर्स लेना जिनका कोरिलेशन (Correlation) 0.5 से ज्यादा है)
correlations = {"feature1":0.8,"feature2":0.2,"feature3":0.6}
strong_features = {n:v for n,v in correlations.items() if v >0.5}                                         
print(strong_features)

# उदाहरण 3: डिक्शनरी की वैल्यूज को ट्रांसफॉर्म करना (Categorical Encoding)
raw_labels = {"user1": "spam", "user2": "ham", "user3": "spam"}
encoded = {u:(1 if m == "spam" else 0) for u,m in raw_labels.items()}
print(encoded)

# उदाहरण 4: दो लिस्ट को मिलाकर डिक्शनरी बनाना (zip के साथ)
columns = ['id','price','area']
values = [101,500000,1200]
row_dict = {col:val for col,val in zip(columns,values)}
print(row_dict)

# उदाहरण 5.
signals = {"Reliance": "Buy", "TCS": "Hold", "Zomato": "Buy"}

# 'Buy' है तो 1 दो, नहीं तो 0 दो
encoded = {company:(1 if stock =="Buy" else 0)for company , stock in signals.items()}
print(encoded)

# उदाहरण 6.
cols = ["Company", "Price", "Sector"]
vals = ["Zomato", 180, "Food Delivery"]

# दोनों लिस्ट को एक साथ लूप करके डिक्शनरी बनाना

looping_dict ={key:value for key,value in zip(cols , vals)}
print(looping_dict)

# उदाहरण 7.
trade_history = {"Trade1": "Loss", "Trade2": "Profit", "Trade3": "Profit"}

# सबसे आखिर वाला (Trade3) पहले प्रिंट होगा
for trade in reversed(trade_history):
    print(trade)
    
```
