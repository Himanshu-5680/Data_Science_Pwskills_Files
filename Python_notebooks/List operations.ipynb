# 1. बेसिक और एक्सेस (Basics & Access)

```python
# [] : creating a list (age of customers)
ages = [25,30,22,35,45]

# list() : data ko list banana (character extraction) 
cols = list("XYZ")
print(cols)

# [i] indexing.
print(ages[0])
print(ages[-1])

# [start:stop:jump] : slicing (Train/Test split jaisa)
print(ages[0:4:2])

# len() : total rows counting.
print(len(ages))

# in / not in : missing ya tageted values check karna (Anomaly detection.)
print(22 in ages)
print(25 not in ages)
```

# 2. आइटम जोड़ना और हटाना (Add & Remove)

```python
# Adding 
#append(x) : new record last me add karna.
ages.append(50) , ages

# insert (i,x) : missing value imputation (Specific Place par values add karna.)
ages.insert(3,26) , ages

# extend(iterable) : adding a new list in old list.
new_ages = [60,65]
ages.extend(new_ages) , print(ages)

# + Making a new list by adding two lists.
batch1 = [1,2]
batch2 = [3,4]
all_batches = batch1 + batch2
print(all_batches)

# Dummy deta create karna (Data augmentation)
zeros = [0] * 5 
zeros 

# Remove(x) : outlier Hatana.
ages.remove(65) # 65 ko remove karna.
last_record = ages.pop() # last item ko remove karna.
del ages[0] # index 0 wale item ko remove karna.
del ages[0:8] # delete column or a part

ages.clear() # Memory khali karna
```

# 6. लिस्ट के साथ बिल्ट-इन फंक्शन्स

```python
# शेयर मार्केट के डेली रिटर्न्स
returns = [0.05, -0.02, 0.1, -0.05, 0.08]

# min() / max()
print(min(returns))
print(max(returns))

# sum() : total returns
print(sum(returns))

# any() : ek bhi positive value hai ?
print(any([r>0 for r in returns]))

# all() : sarai positive values hai ?
print(all([r>0 for r in returns]))

# enuenumerate() : index and value ek saath (Feature mapping)
features = ['age','income','city']
for i , col in enumerate(features):
    print(f'Column {i} : {col}')

# zip() : दो कॉलम्स को एक साथ लूप करना (Actual vs Predicted Model Error)
y_true = [1 , 0 , 1] 
y_pred = [0.9 , 0.1 , 0.8]
for true , pred in zip(y_true , y_pred):
    print(f"Error: {true-pred}")
```

# 7. एडवांस्ड (Data Scientist's Core Tools)
- List Comprehension

```python
# उदाहरण 1: बेसिक ऑपरेशन (सभी वैल्युएशन में 10% का इज़ाफ़ा करना)
base_valuation = [100, 200, 300]
new_valuation = [i*1.1 for i in base_valuation]
print(new_valuation)

# उदाहरण 2: फिल्टरिंग (सिर्फ उन कस्टमर्स को लेना जिनकी उम्र 18 से ऊपर है)
raw_ages = [15, 22, 17, 35, 12, 40]
adults = [age for age in raw_ages if age >= 18]
print(adults)

# उदाहरण 3: टेक्स्ट क्लिनिंग (कॉलम के नामों से स्पेस हटाना और छोटे अक्षरों में बदलना)
cols = ["  Age ", "Income", "  CITY "]
clean_cols = [c.strip().lower() for c in cols
print(clean_cols)
```

- filter() (कंडीशन लगाकर डेटा छांटना)

```python
# उदाहरण 1: डेटा में से Missing/NaN/None वैल्यू हटाना
dirty_data = [10, None, 20, None, 30]
clean_data = list(filter(lambda x: x is not None , dirty_data))
print(clean_data)

# उदाहरण 2: सिर्फ प्रॉफिट वाले दिन (Positive Returns) निकालना
pnl = [-100, 50, -20, 200, 0]
profitable_days = list(filter(lambda x: x > 0 , pnl))
print(profitable_days)

# उदाहरण 3: सिर्फ उन कैटेगरी को रखना जिनमें 'tech' वर्ड हो
categories = ["edtech", "food", "fintech", "retail"]
tech_cats = list(filter(lambda x: 'tech' in x , categories))
print(tech_cats)
```

- map() (पूरी लिस्ट पर एक साथ फंक्शन अप्लाई करना)

```python
# उदाहरण 1: स्ट्रिंग वाले नंबर्स को फ्लोट (Float) में बदलना
str_nums = ["1.5", "2.8", "3.0"]
float_nums = list(map(float , str_nums))
print(float_nums)

# उदाहरण 2: कैटेगोरिकल डेटा को बाइनरी (0 और 1) में एनकोड (Encode) करना
genders = ["M", "F", "M", "M"]
encode = list(map(lambda x : 1 if x =='M' else 0 , genders))
print(encode)


# उदाहरण 3: दो लिस्ट्स के बीच का अंतर निकालना (A/B testing metrics)
group_A_clicks = [10, 20, 30]
group_B_clicks = [12, 18, 35]
difference = list(map(lambda a,b : a-b , group_A_clicks , group_B_clicks))
print(difference)
```

* 1. Outlier Removal (Anomaly Detection)


* समस्या: हमारे पास यूज़र्स की उम्र का डेटा है, जिसमें कुछ गलत एंट्रीज (आउटलायर्स) हैं। `filter()` और `lambda` का इस्तेमाल करके सिर्फ वो उम्र फिल्टर करो जो 18 से 60 के बीच हों।
* सैंपल डेटा:





Python

`ages = [25, 17, 85, 42, -5, 30, 55, 120]`

---

```python
ages = [25, 17, 85, 42, -5, 30, 55, 120]
filtered_ages = list(filter(lambda x : x>18 and x<60, ages))
filtered_ages
```

* समस्या: मशीन लर्निंग मॉडल टेक्स्ट नहीं समझता। `map()` और `lambda` का इस्तेमाल करके चर्न डेटा की लिस्ट में 'Yes' को 1 और 'No' को 0 में बदलो।

Python

`churn_data = ["Yes", "No", "No", "Yes", "No"]`

```python
churn_data = ["Yes", "No", "No", "Yes", "No"]
encoded = list(map(lambda x : 1 if x == "Yes" else 0 , churn_data))
encoded
```

* 3. Min-Max Normalization (Data Scaling)


* समस्या: फीचर्स को स्केल करना ज़रूरी है। लिस्ट कॉम्प्रिहेंशन का इस्तेमाल करके हर वैल्यू को लिस्ट की सबसे बड़ी (max) वैल्यू से डिवाइड करो, ताकि सारा डेटा 0 और 1 की रेंज में आ जाए।
* सैंपल डेटा:





Python

`raw_features = [20, 50, 10, 80, 40]`

---

```python
raw_features = [20, 50, 10, 80, 40]
scale = [i/max(raw_features) for i in raw_features ]
scale
```
