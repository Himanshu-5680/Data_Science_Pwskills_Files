* 2. Handling Multicollinearity (Nested Loop)


* समस्या: एक 2D मैट्रिक्स (लिस्ट के अंदर लिस्ट) है जो फीचर्स का कोरिलेशन दिखा रही है। नेस्टेड लूप (लूप के अंदर लूप) का इस्तेमाल करो और उन पेयर्स के इंडेक्स (i, j) निकालो जिनका कोरिलेशन 0.8 से ज़्यादा है। (ध्यान रहे: i और j बराबर नहीं होने चाहिए)।
* सैंपल डेटा:





Python

`matrix = [[1.0, 0.9, 0.2], [0.9, 1.0, 0.4], [0.2, 0.4, 1.0]]`

---

```python
matrix = [[1.0, 0.9, 0.2], [0.9, 1.0, 0.4], [0.2, 0.4, 1.0]]
for i in range(len(matrix)):
    for j in range(len(matrix)):
        if i!=j and matrix[i][j]> 0.8:
            print(i,j)
```

1. सिंपल लॉजिक (बिल्डिंग का उदाहरण)
इस matrix को एक 3-मंजिला बिल्डिंग समझो।

बाहरी लूप (i): यह तय करता है कि तुम किस मंजिल (Floor) पर हो (0, 1, 2)।

अंदरूनी लूप (j): यह तय करता है कि तुम उस मंजिल के किस कमरे (Room) में जा रहे हो (0, 1, 2)।

matrix[i][j]: मंजिल i के कमरे j के अंदर रखी हुई वैल्यू।

लूप का नियम है: जब तुम किसी एक मंजिल (i) पर जाते हो, तो उसके सारे कमरे (j) चेक करने के बाद ही अगली मंजिल पर जाते हो।

2. बैकएंड का ड्राई रन (कदम दर कदम)
कोड मशीन में कैसे चल रहा है, इसे देखो:

मंजिल 0 पर (i = 0):

कमरा 0 (j = 0): matrix[0][0] वैल्यू 1.0 है। शर्त i != j (0 != 0) फेल हो गई। (इग्नोर)

कमरा 1 (j = 1): matrix[0][1] वैल्यू 0.9 है। 0 != 1 (True) और 0.9 > 0.8 (True)। (प्रिंट करेगा: 0 1)

कमरा 2 (j = 2): matrix[0][2] वैल्यू 0.2 है। शर्त 0.2 > 0.8 फेल हो गई। (इग्नोर)

मंजिल 1 पर (i = 1):

कमरा 0 (j = 0): matrix[1][0] वैल्यू 0.9 है। 1 != 0 (True) और 0.9 > 0.8 (True)। (प्रिंट करेगा: 1 0)

कमरा 1 (j = 1): matrix[1][1] वैल्यू 1.0 है। शर्त 1 != 1 फेल हो गई। (इग्नोर)

कमरा 2 (j = 2): matrix[1][2] वैल्यू 0.4 है। शर्त 0.4 > 0.8 फेल हो गई। (इग्नोर)

(यही काम मंजिल 2 के लिए भी होगा)

3. i != j की शर्त क्यों लगाई?
डेटा साइंस में मैट्रिक्स के बीच वाले हिस्से (Diagonal) में हमेशा 1.0 होता है, क्योंकि वो फीचर का खुद के साथ ही रिलेशन बताता है (जैसे Age का Age से)। हमें वो नहीं देखना है।

हम ढूँढ रहे हैं कि क्या कोई दो अलग-अलग फीचर्स आपस में बहुत ज़्यादा जुड़े हैं? अलग-अलग फीचर्स को पक्का करने के लिए ही हमने i != j (मंजिल और कमरा अलग हों) की शर्त लगाई है।
