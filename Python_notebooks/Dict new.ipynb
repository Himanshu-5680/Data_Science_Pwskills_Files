### 1. Accessing (डेटा निकालना)

* **Visual Analogy:** स्कूल का टिफिन बॉक्स!  तुम्हारे पास एक टिफिन बॉक्स है जिस पर नाम (Key) लिखे हैं: "समोसा", "लड्डू"। तुम बस नाम बोलते हो और अंदर से वो चीज़ (Value) मिल जाती है।
* **Core Logic:** बस 'Key' (नाम) का इस्तेमाल करो। अगर वो डब्बे में है, तो चीज़ मिल जाएगी। `.get()` का इस्तेमाल करने से अगर डब्बा खाली भी हो, तो एरर नहीं आता, बस "कुछ नहीं" मिलता है।
* **Exact Syntax:**

```python
tiffin = {"Samosa": 2, "Laddu": 3}

# सिंपल तरीका
print(tiffin["Samosa"])  # output = 2

# सेफ तरीका (अगर जलेबी ना हो तो 0 दे दो)
print(tiffin.get("Jalebi", 0)) # output = 0

```

### 2. Updating (डेटा मिलाना / बदलना)

* **Visual Analogy:** दिवाली की मिठाइयों का डिब्बा! पहले से डिब्बे में काजू कतली है। अब कोई मेहमान आता है और नई मिठाई (गुलाब जामुन) दे जाता है, या पुरानी काजू कतली की जगह नई ताज़ा काजू कतली रख देता है।
* **Core Logic:** तुम नई 'Key' और 'Value' देते हो। अगर नाम पहले से है, तो पुरानी चीज़ बदल जाती है। अगर नहीं है, तो नई चीज़ जुड़ जाती है।
* **Exact Syntax:**

```python
sweets = {"Kaju Katli": 10}

sweets["Gulab Jamun"] = 5  # नई मिठाई जोड़ी
sweets["Kaju Katli"] = 15  # पुरानी मिठाई अपडेट की

print(sweets)

# output = {"Kaju Katli":15,"Gulab Jamun"5:}

```

### 3. Dictionary Comprehension (जादुई मशीन)

* **Visual Analogy:** एक जादुई मशीन!  तुम मशीन में 10 रुपये डालते हो और वो उसे तुरंत 100 रुपये में बदलकर एक नए डिब्बे में पैक कर देती है।
* **Core Logic:** एक ही लाइन में पुरानी डिक्शनरी को लूप करके नई डिक्शनरी बनाना। बिना लंबी लाइनें लिखे डेटा को तेज़ी से बदलना।
* **Exact Syntax:**

```python
numbers = {"A": 1, "B": 2}

# जादुई मशीन: सबको 10 से गुणा कर दो
magic_box = {k: v * 10 for k, v in numbers.items()}

print(magic_box)
# output = {"A": 10, "B": 20}

```

### 4. Filtering (छलनी / फिल्टर)

* **Visual Analogy:** चाय की छलनी!  तुम डिक्शनरी की चाय को छलनी से निकालते हो। सिर्फ वो चीज़ें (जैसे अच्छे मार्क्स) कप में जाती हैं जो तुम्हारी कंडीशन को पास करती हैं, बाकी कचरा बाहर!
* **Core Logic:** डिक्शनरी कॉम्प्रिहेंशन में बस एक `if` कंडीशन लगा दो। जो कंडीशन पास करेगा, वही नई डिक्शनरी में जाएगा।
* **Exact Syntax:**

```python
scores = {"Aman": 40, "Rahul": 90, "Riya": 85}

# छलनी: सिर्फ 80 से ज्यादा वाले पास होंगे
toppers = {k: v for k, v in scores.items() if v > 80}

print(toppers)
#output = {"Rahul": 90, "Riya": 85}

```

### 5. Zipping (चेन बंद करना)

* **Visual Analogy:** स्कूल बैग की चेन (Zip)!  एक तरफ हुक हैं (Keys), दूसरी तरफ कुंडे (Values)। जैसे ही चेन बंद करते हो, दोनों आपस में जुड़कर एक जोड़ी बन जाते हैं।
* **Core Logic:** दो अलग-अलग लिस्ट्स लो। `zip()` उन्हें एक साथ जोड़ता है, और `dict()` उसे डिक्शनरी बना देता है।
* **Exact Syntax:**

```python
items = ["Bat", "Ball"]
prices = [500, 100]

# दोनों को ज़िप करके डिक्शनरी बनाओ
shop = dict(zip(items, prices))

print(shop)

```

### 6. Grouping (एक जगह इकट्ठा करना)

* **Visual Analogy:** क्लास में बच्चों के हाउस (Red House, Blue House)।  तुम हर बच्चे को उसके रंग की टी-शर्ट के हिसाब से अलग-अलग कमरों (Lists) में भेज रहे हो।
* **Core Logic:** `.setdefault()` का जादू। यह चेक करता है कि "कमरा" बना है या नहीं। अगर नहीं, तो खाली कमरा `[]` बनाता है और फिर उसमें नाम (`append`) डाल देता है।
* **Exact Syntax:**

```python
students = [("Red", "Aman"), ("Blue", "Riya"), ("Red", "Rahul")]
houses = {}

for color, name in students:
    houses.setdefault(color, []).append(name)

print(houses)

```

### 7. Unpacking (सामान बाहर निकालना)

* **Visual Analogy:** जादुई सूटकेस! जब तुम टूर पर जाते हो, तो सूटकेस खोलते ही (``) सारा सामान अपने आप सही अलमारी में सज जाता है।
* **Core Logic:** जब डिक्शनरी के आगे `` लगाते हैं, तो उसकी सारी Keys और Values खुल जाती हैं और किसी फंक्शन को सीधे इनपुट्स की तरह मिल जाती हैं।
* **Exact Syntax:**

```python
def make_profile(name, age):
    print(f"{name} is {age} years old.")

user = {"name": "Himanshu", "age": 22}

# सूटकेस खोल दिया (Unpack कर दिया)
make_profile(**user) 

```
