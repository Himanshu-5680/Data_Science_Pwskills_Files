### 1. बेसिक `for` लूप

**सिंटैक्स:**

```python
for variable in iterable:
    # काम (Code)

```

**कैसे काम करता है:** लूप `iterable` (जैसे लिस्ट) के पहले आइटम को उठाता है और `variable` में डाल देता है, फिर अंदर का कोड चलता है। इसके बाद वह दूसरा आइटम उठाता है। यह तब तक चलता है जब तक लिस्ट खाली न हो जाए।

### 2. `range()` के साथ

**सिंटैक्स:**

```python
for i in range(start, stop, step):
    # काम

```

**कैसे काम करता है:** यह सीधे नंबर्स पैदा करता है। अगर `range(5)` लिखा, तो मशीन अपने आप `[0, 1, 2, 3, 4]` की लिस्ट बना लेगी और लूप 5 बार चलेगा।

### 3. लूप कंट्रोल (`break` और `continue`)

**सिंटैक्स और काम:**

```python
for num in [10, 20, 30, 40]:
    if num == 20:
        continue  # 20 वाले चक्कर को यहीं छोड़ दो, सीधे 30 पर जाओ।
    
    if num == 40:
        break     # 40 आ गया? लूप को तुरंत तोड़ दो और पूरी तरह बाहर आ जाओ।

```

### 4. हेल्पर्स (`enumerate` और `zip`)

**`enumerate` सिंटैक्स:**

```python
for index, value in enumerate(my_list):

```

**कैसे काम करता है:** यह सिर्फ आइटम (value) नहीं देता, बल्कि उसकी जगह (index) भी साथ देता है। (जैसे: `0, "Apple"`, `1, "Mango"`)।

**`zip` सिंटैक्स:**

```python
for item1, item2 in zip(list1, list2):

```

**कैसे काम करता है:** यह दो लिस्ट्स को एक साथ जोड़कर दोनों में से एक-एक आइटम निकालता है। (लिस्ट 1 का पहला और लिस्ट 2 का पहला आइटम एक साथ)।

### 5. नेस्टेड लूप (Nested Loop)

**सिंटैक्स:**

```python
for i in [1, 2]:          # बाहरी लूप
    for j in ['A', 'B']:  # अंदर का लूप
        print(i, j)

```

**कैसे काम करता है:** बाहरी लूप के **हर एक चक्कर** के लिए, अंदर वाला लूप अपना **पूरा चक्कर** खत्म करता है। (आउटपुट: `1 A`, `1 B`, फिर `2 A`, `2 B`)।

बस दो-तीन और बहुत ज़रूरी चीज़ें बची हैं जो एक डेटा साइंटिस्ट रोज़ इस्तेमाल करता है:

### 1. `for ... else` (पाइथन का खास फीचर)

**सिंटैक्स:**

```python
for num in [1, 2, 3]:
    if num == 5:
        break
else:
    print("Loop poora chala, 5 nahi mila!")

```

**कैसे काम करता है:** `else` ब्लॉक सिर्फ तभी चलता है जब लूप **बिना `break` हुए** पूरा खत्म हो जाए। अगर लूप बीच में टूट (break) गया, तो `else` वाला हिस्सा नहीं चलेगा। (डेटाबेस में कुछ सर्च करते समय यह बहुत काम आता है)।

### 2. List Comprehension (One-Liner Loop)

**सिंटैक्स:**

```python
new_list = [x * 2 for x in old_list if x > 0]

```

**कैसे काम करता है:** यह पूरा `for` लूप, `if` कंडीशन और `append()` का काम एक ही लाइन में कर देता है। डेटा क्लीनिंग और फीचर इंजीनियरिंग में यह सबसे तेज़ तरीका है।

### 3. `itertools` लाइब्रेरी (Advanced Looping)

जब बहुत बड़े डेटा पर कॉम्प्लेक्स लूपिंग (जैसे Permutations, Combinations, या कई लिस्ट को एक साथ जोड़ना) करनी हो, तो हम नॉर्मल लूप की जगह पाइथन की `itertools` लाइब्रेरी का इस्तेमाल करते हैं। यह मेमोरी बचाता है और कोड को बहुत तेज़ कर देता है।

#### 2. लूप कंट्रोल (`break`, `continue`, `pass`) - ⭐️ [Very Important]

````python
# उदाहरण 1 (continue): करप्ट या Missing डेटा को स्किप करना
data_points = [100, None , 200, "Error", 300]
valid_sum = 0
for val in data_points:
    if type(val) != int:
        continue
    valid_sum +=val
print(valid_sum)

# it gives us sum of all integers not None type or string type.
````

````python
# उदाहरण 2 (break): थ्रेशोल्ड (Threshold) पार होने पर लूप रोकना (Early Stopping)
losses = [0.5, 0.4, 0.3, 0.1, 0.05, 0.01]
for ep , loss in enumerate(losses):
    if loss<0.1:
        print(f"Model trained perfectly at epoch {ep} at {loss}. Stopping!")
        break # we got our target so now we don't have to continue the loop.
````

````python
# उदाहरण 3 (pass): बाद में कोड लिखने के लिए स्ट्रक्चर बनाना
for feature in ["age","salary"]:
    # TODO: यहाँ बाद में फीचर स्केलिंग का कोड आएगा
    pass
````

````python
# उदाहरण 4 (continue + break combo): डेटा स्कैनिंग
server_logs = ["OK", "OK", "Timeout", "OK", "FATAL_ERROR", "OK"]
for log in server_logs:
    if log == "OK":
        continue
    if log == "FATAL_ERROR":
        print("System down! Stopping process.")
        break
    print(f"Minor issue detected: {log}")
````

#### 3. हेल्पर्स (`enumerate`, `zip`) - ⭐️ [Very Important]

````python
# उदाहरण 1 (enumerate): इंडेक्स का इस्तेमाल करके ओरिजिनल डेटा अपडेट करना
salaries = [40000, 50000, 60000]
# सभी सैलरी में 10% इंक्रीमें
for i , salary in enumerate(salaries):
    salaries[i] = salary*1.10  # i stands for index or variable.
print(salaries)
````

````python
# उदाहरण 2 (zip): Actual vs Predicted वैल्यू की तुलना (Model Evaluation)
y_true = [1, 0, 1, 1]
y_pred = [1, 0, 0, 1]
errors = 0
for true,pred in zip(y_true,y_pred):
    if true != pred:
        errors +=1
    print(f"Total Errors :", errors)
````

````python
# उदाहरण 3 (zip multiple): 3 अलग-अलग कॉलम्स को एक साथ लूप करना
user_ids = [1, 2]
clicks = [150, 20]
purchases = [3, 0]
for uid , click , purchase in zip(user_ids , clicks , purchases):
    convertion_rate = (purchase / click)*100
    print(f"User {uid} Conversion:{convertion_rate}%")
````

````python
# उदाहरण 4 (enumerate + zip): किस रो (Row) में मॉडल फेल हुआ, उसका इंडेक्स निकालना
y_true = [1, 0, 1, 1]
y_pred = [1, 0, 0, 1]
for index , (true , pred) in enumerate(zip(y_true,y_pred)):
    if true != pred:
        print(f"Misclassification found at row {index}")
````

#### 4. एडवांस्ड (`for...else` और Nested) - ⭐️ [Rare but Powerful]

````python
# उदाहरण 1 (for...else): सर्च ऑपरेशन में फ्लैग (Flag) वेरिएबल से बचना
# डेटाबेस में टारगेट खोजना
target_user = "user_404"
active_users = ["user_1", "user_2", "user_3"]

for user in active_users:
    if user == target_user:
        print("User found")
else:
    # यह लाइन तभी चलेगी जब पूरा लूप खत्म हो जाए और break ना चला हो
    print("User not found in the database. Adding new user...")
````

````python
# उदाहरण 2 (Nested Loop): 2D ग्रिड या इमेजेज/मैट्रिक्स का डेटा पढ़ना (Deep Learning)
correlation_matrix = [
    [1.0, 0.8, 0.2],
    [0.8, 1.0, -0.5],
    [0.2, -0.5, 1.0]
]

# ऐसे फीचर्स खोजना जो एक-दूसरे से बहुत ज्यादा जुड़े हुए हैं (Multicollinearity)

for i in range(len(correlation_matrix)):
    for j in range(len(correlation_matrix)):
        if i != j and correlation_matrix[i][j] > 0.7:
            print(f"High correlation between feature {i} and {j}")
````

### मॉडरेट (Moderate)

---
* 1. Early Stopping (`break`)


* समस्या: मॉडल ट्रेनिंग के दौरान एक लॉस (loss) लिस्ट जनरेट हुई है। लूप चलाओ, और जैसे ही लॉस 0.1 से नीचे जाए, लूप को `break` कर दो और बताओ कि यह किस इंडेक्स पर हुआ।
* सैंपल डेटा:





Python

`loss_values = [0.5, 0.35, 0.2, 0.15, 0.08, 0.05]`

---

````python
loss_values = [0.5, 0.35, 0.2, 0.15, 0.08, 0.05]
for i , loss in enumerate(loss_values):
    if loss < 0.1:
        break
print(f"Loss {loss} is index : {i} par huaa tha.")
````

### एडवांस्ड (Advanced)

---
* 1. Target Search (`for...else`)


* समस्या: ब्लैकलिस्टेड यूज़र्स में एक टारगेट आईडी ("U999") खोजनी है। लूप लगाओ। अगर मिले तो प्रिंट करो "Found" और `break` करो। अगर आईडी न मिले, तो `else` ब्लॉक (लूप के साथ वाला) का इस्तेमाल करके "Not Found" प्रिंट करो।
* सैंपल डेटा:





Python

`blacklisted = ["U101", "U202", "U303", "U404"]`

---

````python
blacklisted = ["U101", "U202", "U303", "U404"]
target_id = "U999"
for i in blacklisted:
    if i == target_id:
        print("Found")
        break
else:
    print("Not Found")
````

---

* 2. Handling Multicollinearity (Nested Loop)


* समस्या: एक 2D मैट्रिक्स (लिस्ट के अंदर लिस्ट) है जो फीचर्स का कोरिलेशन दिखा रही है। नेस्टेड लूप (लूप के अंदर लूप) का इस्तेमाल करो और उन पेयर्स के इंडेक्स (i, j) निकालो जिनका कोरिलेशन 0.8 से ज़्यादा है। (ध्यान रहे: i और j बराबर नहीं होने चाहिए)।
* सैंपल डेटा:





Python

`matrix = [[1.0, 0.9, 0.2], [0.9, 1.0, 0.4], [0.2, 0.4, 1.0]]`

---

````python
matrix = [[1.0, 0.9, 0.2], [0.9, 1.0, 0.4], [0.2, 0.4, 1.0]]
for i in range(len(matrix)):
    for j in range(len(matrix)):
        if i!=j and matrix[i][j]> 0.8:
            print(i,j)
````

---

* 3. Multi-Condition Cleaning (Combo)


* समस्या: सेंसर से डेटा आ रहा है। लूप चलाओ। अगर वैल्यू 0 है, तो उसे `continue` से इग्नोर करो। अगर वैल्यू 999 (क्रैश कोड) है, तो लूप को `break` कर दो। बाकी सारी वैलिड रीडिंग्स को एक नई लिस्ट में डालो।
* सैंपल डेटा:





Python

`sensor_data = [25, 0, 30, 45, 0, 999, 50, 60]`

---

````python
sensor_data = [25, 0, 30, 45, 0, 999, 50, 60]
readings = []
for i in sensor_data:
    if i == 0:
        continue
    elif i == 999:
        break
    else :
        readings.append(i)
print(readings)
````
